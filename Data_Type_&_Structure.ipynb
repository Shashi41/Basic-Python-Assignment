2. Explain the difference between mutable and immutable data types with examples
    - Mutable Data Types:
These are objects whose state can be modified after creation. Examples include:
      - Lists: my_list = [1, 2, 3]; my_list.append(4)
      - Dictionaries: my_dict = {'a': 1}; my_dict['b'] = 2
      - Sets: my_set = {1, 2}; my_set.add(3)

   - Immutable Data Types:
      These are objects whose state cannot be changed after creation. Examples include:
      - Integers: x = 5; # can't change x's value directly
      - Floats: y = 3.14; # can't change y's value directly
      - Strings: my_str = 'hello'; # can't change my_str's characters directly
      - Tuples: my_tuple = (1, 2); # can't add or remove elements

4. Describe how dictionaries store data ?
    - Python dictionaries are highly optimized data structures that store data in key-value pairs. They are based on a concept called a Hash Table, which allows you to retrieve data almost instantaneously, regardless of how large the dictionary is.
      - 1. The Hashing Process
When you assign a value to a key, such as my_dict["name"] = "Alice", Python doesn't just put that entry at the end of a list. Instead, it follows these steps:
            - Hashing
            - Indexing
            - Storage
      - 2. Handling Collisions
Sometimes, two different keys might produce the same hash value or point to the same index. This is known as a collision.      
  
      - 3. Insertion Order (Modern Python)
Since Python 3.7+, dictionaries are officially ordered. While the hashing process determines where data lives in memory, Python maintains a separate, compact array that keeps track of the order in which items were added. This means when you print a dictionary, the items appear in the same order you typed them.

8. What is a hash table, and how does it relate to dictionaries in Python ?
    - A hash table is a data structure that maps "keys" to "values" using a mathematical process called hashing. It is essentially a big array (a contiguous block of memory) where each slot is called a bucket.

      - How it works:
          - Hashing: When you provide a key (like "name"), Python passes it through a hash function—specifically hash()—which converts the key into a large integer.

          - Indexing: That large integer is then "compressed" (usually using a modulo operation like hash % array_size) to fit within the range of the current array's indices.

          - Storage: The value is stored in the bucket at that calculated index.

10. Explain why strings are immutable in Python.
    - Strings are immutable in Python for three key reasons:

       1. **Optimization:** Python uses "string interning" to save memory by pointing identical strings to the same memory address. If strings were mutable, changing one would unexpectedly change all others sharing that space.
        2. **Security:** Strings often store sensitive data like file paths or passwords. Immutability ensures these values remain constant after validation, preventing accidental or malicious modification.
        3. **Hashability:** Because strings cannot change, their hash value (fingerprint) remains constant. This allows them to be used as reliable keys in dictionaries and elements in sets, ensuring fast lookups.

11. What advantages do dictionaries offer over lists for certain tasks ?
    - Dictionaries provide three main advantages over lists:

      1. **Speed:** Dictionaries use hash tables for  lookups. Finding an item in a list of a million elements requires scanning them all (), while a dictionary finds it instantly regardless of size.
      2. **Readability:** Accessing data via meaningful keys (e.g., `user['email']`) is more intuitive than using arbitrary numeric indexes (e.g., `user[2]`), making code self-documenting.
      3. **Efficiency:** Dictionaries handle sparse data better, storing only existing key-value pairs rather than requiring a continuous block of memory like a list.

12. Describe a scenario where using a tuple would be preferable over a list.

13. How do sets handle duplicate values in Python in 100 words ?
    - Sets handle duplicates by automatically discarding any value that is already present. This is possible because sets are built on **hash tables**.

      When you add an element, Python calculates its unique **hash** to determine its "bucket" in memory. If you try to add the same value again, Python generates the same hash, identifies that the bucket is already occupied by that identical object, and simply does nothing.

      This makes sets an efficient tool for deduplication—converting a list to a set instantly removes all repeating elements while maintaining  performance.

14. How does the “in” keyword work differently for lists and dictionaries ?
    - The `in` keyword differs primarily in **efficiency** and **target**:

      1. **Lists (Value Search):** Python performs a **linear search**, scanning every item from start to finish. This is  complexity, meaning the search time grows as the list grows.
      2. **Dictionaries (Key Search):** Python uses **hashing** to jump directly to the key's location. This is  complexity, providing near-instant lookups regardless of size. Note that `in` only checks **keys**; checking `.values()` reverts to a slow linear search.

      Essentially, `in` for lists is a "search," while for dictionaries, it is a "calculation."

15. Can you modify the elements of a tuple? Explain why or why not.
    - No, you cannot modify the elements of a tuple because they are **immutable**. Once a tuple is defined, its size and the references it holds are "frozen" in memory.

      This design ensures **data integrity** and **thread safety**. By preventing changes, Python can optimize memory usage and allow tuples to be used as **dictionary keys**, as their hash value will never change.

      However, there is a catch: if a tuple contains a **mutable** object (like a list), you can modify the *contents* of that list, but you still cannot replace the list itself with a different object.

17. Describe the time complexity of accessing elements in a dictionary.
    - The time complexity for accessing elements in a dictionary is ****, or "constant time." This means that retrieving a value takes the same amount of time regardless of whether the dictionary contains ten items or ten million.

      This efficiency is achieved through **hashing**. When you request a key, Python applies a hash function to turn that key into a unique integer index. It then jumps directly to that specific memory location (bucket) to grab the value, rather than scanning the entire collection sequentially.

      While rare "hash collisions" can slightly slow this down to , Python’s internal optimizations keep access virtually instantaneous in almost all practical scenarios.

18. In what situations are lists preferred over dictionaries.
    - Lists are preferred when **order** and **sequence** are your primary concerns. They are ideal for:

      * **Ordered Data:** When the position of elements matters (e.g., a "To-Do" list or a timeline).
      * **Duplicate Values:** When you need to store multiple identical items (e.g., a shopping cart).
      * **Slicing:** When you need to access chunks of data (e.g., `my_list[0:5]`).
      * **Sorting:** When you need to frequently reorder or sort the collection.

19. Why are dictionaries considered unordered, and how does that affect data retrieval ?
    - Historically, dictionaries were considered **unordered** because they used hash tables to determine storage locations. Since a key's hash value is essentially random, the physical order in memory did not match the order in which items were added. This meant you couldn't rely on index-based retrieval (like `dict[0]`).

      Since Python 3.7+, dictionaries **do** preserve insertion order as an implementation detail. However, they remain "unordered" in principle: you still retrieve data by **key**, not by position. This ensures  lookup speed but makes dictionaries unsuitable for tasks requiring numerical slicing or sorting.
  
20. Explain the difference between a list and a dictionary in terms of data retrieval.
    - Data retrieval differs primarily in **access method** and **efficiency**.

      **Lists** use **positional indexing**. To find an item, you must know its numerical index or perform a **linear search** () by checking every element. This makes retrieval slow as the dataset grows.

      **Dictionaries** use **key-value mapping**. They utilize **hash tables** to jump directly to a value using its unique key. This provides **constant time** () retrieval, meaning lookups are nearly instantaneous regardless of size.

      In short, lists are for sequences where order matters; dictionaries are for high-speed retrieval using labels.
