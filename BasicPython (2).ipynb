10. What is the difference between implicit and explicit type casting ?

11. What is the purpose of conditional statements in Python ?

12. How does the elif statement work?
    - A program executes if...elif...else blocks sequentially:
      - if check: The program first evaluates the expression in the initial if statement.

      - If True: The code inside the if block executes, and the rest of the elif and else blocks are completely skipped.

      - If False: The program moves immediately to the next elif statement and evaluates its expression.

      - Repeat elif: This continues down the chain: if an elif condition is True, its block executes, and the rest are skipped.

      - else (Fallback): If every single if and elif condition in the chain is False, the code inside the final else block (if present) executes.

      - Crucially, only one block of code within the entire if...elif...else structure will ever run.

14. Describe a scenario where a while loop is more suitable than a for loop.

      - You don't know if they will guess it on the first try or the fifth try.

```python
# 4.) Write a program that checks if a given word is a Python keyword.

import keyword

def is_python_keyword(word):
  return keyword.iskeyword(word)

word_to_check_1 = "for"
word_to_check_2 = "my_variable"
word_to_check_3 = "True"

if is_python_keyword(word_to_check_1):
  print(f"'{word_to_check_1}' is a Python keyword.")
else:
  print(f"'{word_to_check_1}' is not a Python keyword.")

if is_python_keyword(word_to_check_2):
  print(f"'{word_to_check_2}' is a Python keyword.")
else:
  print(f"'{word_to_check_2}' is not a Python keyword.")

if is_python_keyword(word_to_check_3):
  print(f"'{word_to_check_3}' is a Python keyword.")
else:
  print(f"'{word_to_check_3}' is not a Python keyword.")
```

```python
# 6.) Write a function to demonstrate the behavior of mutable and immutable arguments.

def modify_elements(val, collection):
    print(f"Inside function (start): val = {val}, collection = {collection}")

    val += 10
    collection.append("new_item")

    print(f"Inside function (end):   val = {val}, collection = {collection}")

# Initial values
my_num = 5
my_list = ["original"]

print(f"Before function: my_num = {my_num}, my_list = {my_list}")
print("-" * 40)

modify_elements(my_num, my_list)

print("-" * 40)
print(f"After function:  my_num = {my_num}, my_list = {my_list}")
```

```python
# 7.) Write a program that performs basic arithmetic operations on two user-input numbers.

# Program to input two numbers and perform basic arithmetic operations

# Take input from the user and convert to a floating-point number
try:
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))

    print("\nResults:")
    # Addition
    print(f"Addition: {num1} + {num2} = {num1 + num2}")
    # Subtraction
    print(f"Subtraction: {num1} - {num2} = {num1 - num2}")
    # Multiplication
    print(f"Multiplication: {num1} * {num2} = {num1 * num2}")
    # Division
    if num2 != 0:
        print(f"Division: {num1} / {num2} = {num1 / num2}")
    else:
        print("Division: Error! Division by zero is not allowed.")
    # Modulus (remainder)
    if num2 != 0:
         print(f"Modulus: {num1} % {num2} = {num1 % num2}")
    else:
        print("Modulus: Error! Modulus by zero is not allowed.")
    # Floor Division
    if num2 != 0:
        print(f"Floor Division: {num1} // {num2} = {num1 // num2}")
    else:
        print("Floor Division: Error! Floor division by zero is not allowed.")
    # Exponentiation
    print(f"Exponentiation: {num1} ** {num2} = {num1 ** num2}")

except ValueError:
    print("Error: Invalid input. Please enter numeric values.")

```

```python
# 8.) Write a program to demonstrate the use of logical operators.

print("Welcome to the Logical Operator Demo.")
print("-" * 30)

# 1. Define variables for demonstration
temperature = 25  # Celsius
is_sunny = True
is_weekend = False
is_cloudy = True

print(f"Current conditions:")
print(f"  Temperature: {temperature}°C")
print(f"  Is it sunny? {is_sunny}")
print(f"  Is it the weekend? {is_weekend}")
print(f"  Is it cloudy? {is_cloudy}")
print("-" * 30)


# 2. Demonstration of the 'and' operator
# The 'and' operator returns True if and only if both conditions are True.

# Condition: We only go to the beach if it's both sunny AND warm (>20°C).
goes_to_beach = (is_sunny and temperature > 20)
print(f"Decision for the beach (sunny AND warm): {goes_to_beach}")
# In this case: True and True -> True


# 3. Demonstration of the 'or' operator
# The 'or' operator returns True if at least one of the conditions is True.

# Condition: We stay home if it's cloudy OR it's the weekend.
stays_home = (is_cloudy or is_weekend)
print(f"Decision to stay home (cloudy OR weekend): {stays_home}")
# In this case: True or False -> True (because it is cloudy)


# 4. Demonstration of the 'not' operator
# The 'not' operator reverses the logical state of its operand.

# Condition: We are 'not' working if it is the weekend.
is_not_working = (not is_weekend)
print(f"Are we not working today? (NOT weekend): {is_not_working}")
# In this case: not False -> True


# 5. Combining logical operators (complex expression)

# Condition: Picnic is only good if it's sunny AND warm, AND it's NOT the weekend
# (perhaps we want the park to be less crowded).

is_picnic_time = (is_sunny and temperature > 20 and not is_weekend)
print(f"\nDecision for a picnic (sunny AND warm AND NOT weekend): {is_picnic_time}")
# In this case: (True and True and not False) -> True and True -> True
```

```python
# 9.) Write a Python program to convert user input from string to integer, float, and boolean types.

def to_integer(user_input):
    """Converts a string to an integer."""
    try:
        # Attempt to cast the input string directly to an integer
        return int(user_input)
    except ValueError:
        # Handle the error if the input cannot be converted to a valid integer
        return f"Error: '{user_input}' cannot be converted to an integer."

def to_float(user_input):
    """Converts a string to a float."""
    try:
        # Attempt to cast the input string directly to a float
        return float(user_input)
    except ValueError:
        # Handle the error if the input cannot be converted to a valid float
        return f"Error: '{user_input}' cannot be converted to a float."

def to_boolean(user_input):
    """Converts a string to a boolean (True/False)."""
    # Normalize input to lowercase for consistent comparison
    normalized_input = user_input.lower().strip()

    if normalized_input in ('true', 'yes', '1'):
        return True
    elif normalized_input in ('false', 'no', '0', ''): # Empty string usually evaluates to False in implicit casts
        return False
    else:
        # For strings that don't clearly map to True or False
        return f"Cannot reliably convert '{user_input}' to a boolean type."

# --- Main part of the program to get user input and display conversions ---
if __name__ == "__main__":
    # 1. Get input from the user
    user_input_str = input("Enter a value to convert: ")

    if not user_input_str.strip():
        print("Input was empty. Cannot proceed with conversions.")
    else:
        print(f"\n--- Results for input: '{user_input_str}' ---")

        # 2. Convert and print results for Integer
        int_result = to_integer(user_input_str)
        print(f"To Integer: {int_result}, Type: {type(int_result) if not isinstance(int_result, str) else 'N/A (Error)'}")

        # 3. Convert and print results for Float
        float_result = to_float(user_input_str)
        print(f"To Float:   {float_result}, Type: {type(float_result) if not isinstance(float_result, str) else 'N/A (Error)'}")

        # 4. Convert and print results for Boolean
        bool_result = to_boolean(user_input_str)
        print(f"To Boolean: {bool_result}, Type: {type(bool_result) if not isinstance(bool_result, str) else 'N/A (Error)'}")

```

```python
# 10.) Write code to demonstrate type casting with list elements.

# Original list of strings that represent numbers
str_list = ['1', '2', '3', '4', '5']
print(f"Original list: {str_list} (Type of element: {type(str_list[0])})")

# Type cast to integers using list comprehension
int_list = [int(item) for item in str_list]

print(f"New list: {int_list} (Type of element: {type(int_list[0])})")

# Example casting floats to integers (truncates decimals)
float_list = [1.5, 2.7, 3.1, 4.9]
int_from_float_list = [int(item) for item in float_list]
print(f"\nOriginal float list: {float_list}")
print(f"Casted to int list: {int_from_float_list}")
```

```python
# 14.) Write a program to reverse a string using a while loop.

def reverse_string(text):
    reversed_text = ""

    index = len(text) - 1

    while index >= 0:
        reversed_text += text[index]
        index -= 1

    return reversed_text

input_str = input("Enter any string : ")
result = reverse_string(input_str)

print(f"Original: {input_str}")
print(f"Reversed: {result}")
```

```python
# 15.) Write a Python program to calculate the factorial of a number provided by the user using a while loop.

try:
    number = int(input("Enter a non-negative integer to calculate its factorial: "))
except ValueError:
    print("Invalid input. Please enter an integer.")
    exit()

factorial = 1
counter = 1

if number < 0:
    print("Factorial does not exist for negative numbers.")
elif number == 0:
    print(f"The factorial of {number} is 1.")
else:

    while counter <= number:
        factorial *= counter
        counter += 1

    print(f"The factorial of {number} is {factorial}.")

```
