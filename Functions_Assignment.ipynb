1. 1. What is the difference between a function and a method in Python?

    - ## 1.Function :- A **function** is a standalone block of code that performs a specific task. It is defined outside of any class and can be called directly by its name.
      * **Scope:** Global or within another function.
      * **Data:** It operates only on the data you explicitly pass to it as arguments.
      * **Definition:** Created using the `def` keyword.

**Example:**

```python
def add_numbers(a, b):
    return a + b

# Calling the function
print(add_numbers(5, 10))

```

---

## 2.Method :-

      - A **method** is essentially a function that "belongs to" an object or a class. It is defined inside a class definition.

        * **Scope:** Limited to the object or class it belongs to.
        * **Data:** It can access and manipulate the data (attributes) stored within the object.
        * **The `self` Parameter:** Methods almost always take a special first argument, usually named `self`, which represents the specific instance of the object calling the method.
        * **Calling:** It is called using "dot notation" (e.g., `object.method()`).

**Example:**

```python
class Calculator:
    def __init__(self, value):
        self.value = value

    def multiply_by(self, factor):  # This is a method
        return self.value * factor

# Creating an object (instance)
my_calc = Calculator(10)

# Calling the method on the object
print(my_calc.multiply_by(5))


2. Explain the concept of function arguments and parameters in Python.
    - In Python, **parameters** are the placeholders defined in a function's signature (e.g., `def func(x):`), while **arguments** are the actual values you pass in when calling it (e.g., `func(10)`).

      Python supports four main argument types:

      1. **Positional:** Matched by their order.
      2. **Keyword:** Matched by parameter name, allowing for any order.
      3. **Default:** Parameters with pre-assigned values used if no argument is provided.
      4. **Variable-length:** `*args` collects extra positional arguments into a tuple, while `**kwargs` collects extra keyword arguments into a dictionary.

3. What are the different ways to define and call a function in Python?
    - Python offers three primary ways to define and call functions:

    1. **Standard Functions:** Defined using `def`, these are named, reusable blocks used for complex logic. Call them by name: `my_func()`.
    2. **Lambda Functions:** Anonymous, one-line functions defined with `lambda`. Ideal for quick operations or as arguments to functions like `map()` or `filter()`. Call them via variable name: `add(1, 2)`.
    3. **Methods:** Functions defined inside a class to define object behavior. They require an instance and are called using dot notation: `object.method()`.

      Each type serves a specific purpose, from structured programming to concise, functional-style data processing.

4. What is the purpose of the `return` statement in a Python function?
    - The return statement serves two vital roles in a Python function: it terminates the function’s execution and sends a value back to the caller.

    1. Passing Data Back
      Without a return statement, a function performs a task but "forgets" the result once it finishes. return allows you to capture that result in a variable for later use.


def add(a, b):
    return a + b  # Sends the sum back

  result = add(5, 3) # 'result' now holds the value 8
      2. Immediate Exit
        When Python hits a return line, it exits the function instantly. Any code written inside the function after that line will never run. This is often used for "guard clauses" to exit early if a certain condition is met.

      3. The Default: None
       If you don't include a return statement (or use a bare return), Python automatically returns None. This is why functions like print() return None—they display text to the screen but don't pass a data value back to the code.

5. What are iterators in Python and how do they differ from iterables?
    - An **iterable** is any object you can loop over, such as a list, string, or dictionary. It contains data but doesn't track its own position in the loop. Technically, an iterable implements the `__iter__()` method.

      An **iterator** is the "stateful" object that performs the actual looping. It tracks the current position and produced values one-by-one using the `next()` function (via the `__next__()` method). When no items remain, it raises a `StopIteration` error.

      In short: An **iterable** is the container (the book), while the **iterator** is the bookmark tracking your progress.

6. Explain the concept of generators in Python and how they are defined.
    - A **generator** is a special type of iterator that produces values on-the-fly rather than storing them in memory. Unlike standard functions that use `return` to send a single value and terminate, generators use the **`yield`** keyword.

      When a generator is called, it doesn't run the code immediately; instead, it returns a generator object. Each time `next()` is called, the function executes until it hits `yield`, "pausing" its state to be resumed later.

      This makes generators highly **memory-efficient** for processing large datasets, as they calculate one item at a time (lazy evaluation).

7. What are the advantages of using generators over regular functions?
    - The primary advantage of **generators** is **memory efficiency**. While a regular function returns a complete list—storing every element in RAM simultaneously—a generator yields one item at a time. This allows you to process massive datasets or infinite sequences that would otherwise crash your system.

      Generators also use **lazy evaluation**, meaning computation only happens when a value is requested. This improves performance by avoiding unnecessary calculations. Additionally, they produce cleaner code for complex data streams, as the function maintains its local state automatically between yields, acting as a built-in state machine.

8. What is a lambda function in Python and when is it typically used?
    - A **lambda function** is a small, anonymous function defined with the `lambda` keyword rather than `def`. It is restricted to a single expression and automatically returns the result. The syntax is: `lambda arguments: expression`.

      They are typically used for **short-term, throwaway logic** where naming a full function would be overkill. Common use cases include:

        * **Sorting:** Using the `key` parameter in `sort()` or `sorted()`.
        * **Functional Programming:** Inside `map()`, `filter()`, or `reduce()` to transform data.
        * **Callbacks:** Defining quick event-handling logic in GUI frameworks.

      They prioritize conciseness for simple operations.

9. Explain the purpose and usage of the `map()` function in Python.
    - The `map()` function applies a specific function to every item in an iterable (like a list) without an explicit loop. Its syntax is `map(function, iterable)`.

      Instead of producing a list immediately, `map()` returns an **iterator**. This "lazy evaluation" saves memory because values are only computed as you need them. It is commonly paired with **lambda functions** for quick data transformations, such as converting a list of strings to integers: `map(int, ["1", "2"])`. While powerful and memory-efficient, many developers use list comprehensions as a more readable alternative for simple tasks.

10. What is the difference between `map()`, `reduce()`, and `filter()`    functions in Python?
    - These three functions are the core of Python’s functional programming toolkit, each serving a distinct data-processing purpose:

      * **`map()`**: **Transforms** every item in an iterable. It applies a function to each element and returns an iterator of the results (e.g., doubling every number).
      * **`filter()`**: **Selects** items. It applies a boolean test to each element and only keeps those that return `True` (e.g., keeping only even numbers).
      * **`reduce()`**: **Consolidates** items. Imported from `functools`, it repeatedly applies a function to pairs of elements to reduce the entire list to a **single value** (e.g., calculating the product of all numbers).

````python
# 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

def sum_even_numbers(numbers):
    """
    Calculates the sum of all even numbers in a list.
    """
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total

# Example usage:
my_list = [1, 2, 3, 4, 5, 6]
print(sum_even_numbers(my_list))  # Output: 12 (2 + 4 + 6)

def sum_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)
````

````python
# 5.Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

class FibonacciIterator:
    def __init__(self, limit):
        self.limit = limit    # Maximum number of terms
        self.count = 0        # Current term counter
        self.a, self.b = 0, 1 # Starting values of Fibonacci

    def __iter__(self):
        # An iterator must return itself
        return self

    def __next__(self):
        if self.count >= self.limit:
            # StopIteration tells the loop to finish
            raise StopIteration

        # Calculate the next term
        current_value = self.a
        self.a, self.b = self.b, self.a + self.b

        self.count += 1
        return current_value

# Usage:
fib = FibonacciIterator(8)
for num in fib:
    print(num, end=" ")
# Output: 0 1 1 2 3 5 8 13
````

````python
# 6.Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(max_exponent):
    """
    Yields powers of 2 from 2^0 up to 2^max_exponent.
    """
    exponent = 0
    while exponent <= max_exponent:
        yield 2 ** exponent
        exponent += 1

# Usage:
for val in powers_of_two(5):
    print(val, end=" ")
# Output: 1 2 4 8 16 32
````

````python
# 7.Implement a generator function that reads a file line by line and yields each line as a string.

def read_file_line_by_line(file_path):
    """
    Generator that reads a file and yields one line at a time.
    """
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            for line in file:
                # .strip() removes the trailing newline character '\n'
                yield line.strip()
    except FileNotFoundError:
        print("Error: The file was not found.")

# Usage:
# for line in read_file_line_by_line("data.txt"):
#     print(line)
````

````python
# 8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

# A list of tuples: (Item, Price)
data = [("Apple", 50), ("Banana", 20), ("Cherry", 80), ("Date", 40)]

# Sort the list based on the second element (index 1)
# x represents each tuple, and x[1] is the value we are sorting by
data.sort(key=lambda x: x[1])

print(data)
# Output: [('Banana', 20), ('Date', 40), ('Apple', 50), ('Cherry', 80)]
````

````python
# 9.Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

# List of temperatures in Celsius
celsius_temps = [0, 10, 20, 30, 37, 100]

# Use map with a lambda function to perform the conversion
# The lambda takes 'c' (celsius) and returns 'f' (fahrenheit)
fahrenheit_temps = map(lambda c: (c * 9/5) + 32, celsius_temps)

# Convert the map object to a list to display the results
fahrenheit_list = list(fahrenheit_temps)

print(f"Celsius:    {celsius_temps}")
print(f"Fahrenheit: {fahrenheit_list}")
````

````python
# 10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(input_string):
    vowels = "aeiouAEIOU"


    result_iterator = filter(lambda char: char not in vowels, input_string)


    return "".join(result_iterator)

# Example usage:
text = "Hello, how are you today?"
cleaned_text = remove_vowels(text)

print(f"Original: {text}")
print(f"Result:   {cleaned_text}")
# Output: Hll, hw r y tdy?
````

````python
# 11. Write a Python program, which returns a list with 2-tuples. Each tuple consists of the order number and the
# product of the price per item and the quantity. The product should be increased by 10,- € if the value of the
# order is smaller than 100,00 €. Write a Python program using lambda and map.

# The input data as a list of sublists
orders = [
    [34587, "Learning Python, Mark Lutz", 4, 40.95],
    [98762, "Programming Python, Mark Lutz", 5, 56.80],
    [77226, "Head First Python, Paul Barry", 3, 32.95],
    [88112, "Einführung in Python3, Bernd Klein", 3, 24.99]
]

# The lambda logic:
# 1. Calculate total: order[2] * order[3]
# 2. Check if total < 100. If so, add 10.
# 3. Return a tuple: (order[0], final_total)

result = list(map(lambda x: (x[0], x[2] * x[3] if x[2] * x[3] >= 100 else x[2] * x[3] + 10), orders))

# Displaying the results formatted
print("Order Number | Total Price")
print("-" * 27)
for order_num, total in result:
    print(f"{order_num:<12} | {total:>10.2f} €")
````
